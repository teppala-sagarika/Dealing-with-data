```python
#series

s=pd.Series([1,2,3,4,5]) #index == label
print(s)
print(type(s))

#access using idx
print(s[2])
```

```python
s=pd.Series([1,2,3,4,5],index=["Adam","Eve","Raj","Sag","Moki"]) #index != labels
print(s)

#access using labels
print(s["Adam"])
print(s["Raj"])

#access using idx
print(s[2]) #iloc needed

print(s.index) #all labels
```

```python
#Series properties 

#homogenous values
s1=pd.Series([1,2,3,4,5])
s2=pd.Series([10,20,30,40,50])

#vectorization
print(s1+s2)

#mutable values, immutable size
changed_s1=s1.drop(0)

print(s1)
print(changed_s1)
```

```python
#dataframe 

#dictionary of lists 
info = {
    "Name":["Alice","Bob","Raj","Sag"],
    "Age":[10,16,18,24]
}

df=pd.DataFrame(info)

print(df,type(df))

#labels
print(df.index)
print(df.columns)

#list of lists 

import numpy as np

arr=np.array([[1,2,3],[4,5,6]])
df2=pd.DataFrame(arr,columns=["A","B","C"])

print(df2)
```
